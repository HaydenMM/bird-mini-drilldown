```python
import plotly.graph_objects as go
import json

# Replace this with your actual data
with open("/Users/hmmoore/Desktop/mini_dev/evaluation/path_eval_log_with_questions_2.json", "r") as f:
    data = json.load(f)
```

```python
fig = go.Figure()
y_offset = 0

for path_entry in data:
    path = path_entry["path"]
    x_vals = list(range(len(path)))
    y_vals = [y_offset] * len(path)
    colors = ['green' if step['correct'] == 1 else 'red' for step in path]

    def truncate(text, max_len=220):
        return text if len(text) <= max_len else text[:max_len] + "..."
    hover_texts = []
    for step in path:
        tmp_str = f"QID: {step['question_id']}<br>" + f"<b>Difficulty:</b> {step['difficulty']}<br>" + f"<b>Correct:</b> {'✅' if step['correct'] else '❌'}<br><br>" + f"<b>Question:</b> {step['question']}<br><br>"
        try:
            for hall in step['hallucination']:
                tmp_str += f"<b>Hallucination:</b> {hall}<br><br>"
                
        except:
            tmp_str += f"<b>Hallucination:</b> <br><br>"
        
        hover_texts.append(tmp_str)

    
    # Line
    fig.add_trace(go.Scatter(
        x=x_vals,
        y=y_vals,
        mode='lines+markers',
        marker=dict(color=colors, size=10),
        line=dict(color='gray', width=2),
        text=hover_texts,
        hoverinfo='text',
        showlegend=False
    ))

    y_offset += 1  # stack vertically

# Layout config
fig.update_layout(
    title="SQL Query Path Failures",
    xaxis_title="Query Step in Path",
    yaxis_title="Query Path Index",
    yaxis=dict(showticklabels=False),
    template="plotly_white",
    height=400 + 50 * y_offset,
    margin=dict(l=20, r=20, t=40, b=20),
)

fig.show()
```

```python
def plot_difficulty_block(difficulty_level):
    fig = go.Figure()
    y_offset = 0

    def truncate(text, max_len=220):
        return text if len(text) <= max_len else text[:max_len] + "..."

    total = 0
    correct = 0

    for entry in data:
        path = entry["path"]
        if not path or path[0]["difficulty"] != difficulty_level:
            continue

        total += len(path)
        for step in path:
            if step['correct'] == None:
                step['correct'] = 0
        correct += sum(step['correct'] for step in path)

        x_vals = list(range(len(path)))
        y_vals = [y_offset] * len(path)
        colors = ['green' if step['correct'] == 1 else 'red' for step in path]
        hover_texts = [
            f"QID: {step['question_id']}<br>"
            f"<b>Difficulty:</b> {step['difficulty']}<br>"
            f"<b>Correct:</b> {'✅' if step['correct'] else '❌'}<br><br>"
            f"<b>GT:</b> {truncate(step['gt_query'])}<br><br>"
            f"<b>Pred:</b> {truncate(step['predicted_query'])}"
            for step in path
        ]

        fig.add_trace(go.Scatter(
            x=x_vals,
            y=y_vals,
            mode='lines+markers',
            marker=dict(color=colors, size=10),
            line=dict(color='gray', width=2),
            text=hover_texts,
            hoverinfo='text',
            showlegend=False
        ))

        y_offset += 1

    accuracy = (correct / total * 100) if total > 0 else 0

    fig.update_layout(
        title=f"{difficulty_level.capitalize()} Difficulty – Accuracy: {accuracy:.1f}% ({correct}/{total})",
        xaxis_title="Query Step in Path",
        yaxis_title="Query Path Index",
        yaxis=dict(showticklabels=False),
        template="plotly_white",
        height=400 + 50 * y_offset,
        margin=dict(l=20, r=20, t=40, b=20),
    )
    fig.show()

# Example usage:
plot_difficulty_block("simple")  # You can also try "moderate" or "hard"
```
